## Pandas Visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv(r"C:\Users\kaifi\OneDrive\ドキュメント\Safa\Pandas\Datasets\Ice Cream Ratings.csv")
df
```

```python
df=df.set_index("Date")   # Seting Date column as index
df
```

## Line Chart

```python
print(plt.style.available)
plt.style.use('seaborn-v0_8-dark-palette')    # Optional 
```

```python
df.plot(kind= 'line')   # by default line plot
```

```python
df.plot(kind= 'line', subplots= True) # give plotting columnwise
```

```python
df.plot(kind= 'line' , title= 'Ice-cream Ratings' , xlabel= ' Daily Date-wise Rating' , ylabel = 'Ratings') 
```

## Bar Plot

```python
df.plot(kind = 'bar')
```

```python
df.plot.bar (stacked =True)
```

```python
df['Flavor Rating'].plot.barh(stacked=True)
```

## Scatter Plot

```python
df.plot.scatter( x = 'Texture Rating' , y='Overall Rating' , s=100 , c='Green')  # s=size of the points
                                                                                 # c= color of the points (for more info select shift+Tab inside the code
```

## Histogram  

```python
df.plot.hist(bins=10)  #bin means a range/group of values it tell python how many groups(ranges) you want to distribute the values into
```

## Boxplot

```python
df.plot.box( grid =True)
```

## Area plot

```python
df.plot.area(figsize = (10 , 5))
```

## Pie Chart

```python
df.plot.pie(y = 'Overall Rating', figsize=(10,6))
```
