## Data Cleaning in Pandas 

```python
import pandas as pd
```

```python
df = pd.read_excel(r"C:\Users\kaifi\OneDrive\ドキュメント\Safa\Pandas\Datasets\Customer Call List.xlsx")
df
```

## Remove Duplicates

```python
df.duplicated().any()   # if duplicate exits it will give True else False
```

```python
df.duplicated().sum()  # count of duplicates
```

```python
df[df.duplicated()]  # display duplicate values 
```

```python
df = df.drop_duplicates()          # we are using df = so that the changes are stored in df 
df  
```

## Remove unnecessary column 

```python
df = df.drop(columns = 'Not_Useful_Column')
df
```

## Strip

```python
# to clean the last_name column
```

```python
df['Last_Name'] =df['Last_Name'].str.strip('/._')    #lstrip will remove from left , rstrip will remove from right but we have on both sides so will use strip
df                                     # we want to remove / , . , _ 
```

## Replace

```python
## Clean Phone Number
```

```python
df['Phone_Number']=df['Phone_Number'].str.replace('[^a-zA-Z0-9]','',regex =True)    # regex = Regular Expression (Search pattern or text)
df
```

```python
df['Phone_Number']=df['Phone_Number'].apply(lambda x : str(x))

df['Phone_Number']=df['Phone_Number'].apply(lambda x : x[0:3]+  '-' + x[3:6] + '-' + x[6:10])
df
```

```python
df['Phone_Number']=df['Phone_Number'].str.replace('nan--',' ')
df['Phone_Number']=df['Phone_Number'].str.replace('Na--',' ')
df
```

## Strip

```python
df[['Street_Address', 'State' ,'Zipcode']]=df['Address'].str.split(pat =',' ,n=2, expand = True)  # pat = pattern use cooma to split 
df                                                                                                # n = how many times should it split 
                                                                                                  # expand = put the split result into separate column 
```

```python
df['Paying Customer']=df['Paying Customer'].str.replace('Yes' , 'Y')
df['Paying Customer']=df['Paying Customer'].str.replace('No' , 'N')
df
```

```python
df['Do_Not_Contact']=df['Do_Not_Contact'].str.replace('Yes' , 'Y')
df['Do_Not_Contact']=df['Do_Not_Contact'].str.replace('No' , 'N')
df
```

## Fillna

```python
df=df.fillna(' ')
df
```

```python
## in do not column the customer dosn't want to be contacted so will remove that row 


for x in df.index :
    if df.loc[x , 'Do_Not_Contact'] == 'Y':
       df.drop(x , inplace =True)
df
     
  
```

```python
df['Do_Not_Contact']=df['Do_Not_Contact'].str.replace(' ' , 'N')
df
```

## Drop

```python
# if their is no phone number their is no use to keep it so get rid of it 

for x in df.index:
    if df.loc[x , "Phone_Number"] == ' ':
        df.drop (x , inplace= True)
df


# Another way to drop null values
#df=df.dropna(subset= 'Phone_Number' , inplace =True)
#df
```
