## Group by and Aggregating

```python
# group by will group together the values in a column a display them all in the same row and this allowed you to use aggregate functuions on those columns 
```

```python
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\kaifi\OneDrive\ドキュメント\Safa\Pandas\Datasets\Flavors.csv")
df
```

```python
group_by_frame = df.groupby("Base Flavor")
```

```python
group_by_frame.mean(numeric_only = True)
```

```python
df.groupby("Base Flavor").mean(numeric_only = True) #Mean
```

```python
df.groupby("Base Flavor").count()   # Distinct Values
```

```python
df.groupby("Base Flavor").min()  # Minimum
```

```python
df.groupby("Base Flavor").max()       # Maximum
```

```python
df.groupby("Base Flavor").sum(numeric_only =True)   # sum cant work on str only numeric same as mean
```

```python
df.groupby("Base Flavor").agg({'Flavor Rating' :['mean' , 'min' , 'max' ,'sum'], 'Texture Rating': ['mean' , 'min' , 'max' ,'sum']})
```

```python
df
```

```python
df.groupby(['Base Flavor' ,'Liked']).mean(numeric_only =True)
```

```python
df.groupby(['Base Flavor' ,'Liked']).agg({'Flavor Rating' :['mean' , 'min' , 'max' ,'sum']})
```
